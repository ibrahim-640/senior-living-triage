```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("../data/processed/emails_clean.csv")
df["text"] = df["subject"].fillna("") + " " + df["body"].fillna("")
print(df.shape)
df[["label", "text"]].head(3)
```

```python
train, temp = train_test_split(df, test_size=0.30, stratify=df["label"], random_state=42)
val, test = train_test_split(temp, test_size=0.50, stratify=temp["label"], random_state=42)

print(len(train), len(val), len(test))
```

```python
for name, part in [("train", train), ("val", val), ("test", test)]:
    print(name)
    print(part["label"].value_counts(normalize=True).round(3))
    print()

# No email should appear in two sets. Both lines should print set()
print(set(train["email_id"]) & set(val["email_id"]))
print(set(train["email_id"]) & set(test["email_id"]))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(train["text"], train["label"])
print("Validation accuracy:", round(accuracy_score(val["label"], dummy.predict(val["text"])), 3))
```
